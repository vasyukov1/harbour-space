```python
import numpy as np

def gram_schmidt(vectors):
    basis = []

    for v in vectors:
        v = np.array(v, dtype=float)

        for u in basis:
            v -= np.dot(u, v) * u

        norm = np.linalg.norm(v)

        if norm > 1e-10:
            v /= norm
            basis.append(v)

    return basis

vectors = [
    [1, 2, 3],
    [1, 5, 1],
    [4, 1, 0]
]

result = gram_schmidt(vectors)
for v in result:
    print(v)

print("\nWith norm:")
for v in result:
    print(np.linalg.norm(v))
```
