```python
import numpy as np

def rotate_vector(vector, angle):
    r = np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle), np.cos(angle)]
    ])

    return np.dot(r, vector)


vector = np.array([1, 2])
angle = np.pi / 4

rotated_vector = rotate_vector(vector, angle)
print(rotated_vector)
```
