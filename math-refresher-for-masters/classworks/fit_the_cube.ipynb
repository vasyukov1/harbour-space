```python
import numpy as np
import matplotlib.pyplot as plt

# Set seed
np.random.seed(42)

def solve(a, b, m):
    x = np.linspace(a, b, m)
    y = 2 + 3*x - x**2 + 3*x**3 + np.random.normal(0, noise, m)

    # Build matrix B
    B = np.array([np.ones(m), x, x**2, x**3]).T

    # Solve
    # c = B^T * y
    c = np.linalg.solve(B.T @ B, B.T @ y)

    # Polynomial
    def p(x):
        return c[0] + c[1]*x + c[2]*x**2 + c[3]*x**3

    # Plot 1
    x_plot = np.linspace(a, b, 200)

    plt.scatter(x, y, label="Noisy points")
    plt.plot(x_plot, p(x_plot), color="red", label="Polynomial fit")
    plt.legend()
    plt.grid()
    plt.show()

    print("Coefficients:", c)

# Parameters
m = 20          # Count of points
a, b = -5, 5    # Interval
noise = 5

solve(a, b, m)
```
