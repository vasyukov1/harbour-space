```python
import numpy as np

def kaczmarz(x, w, b):
    return x - w * (b + x @ w) / (w @ w)

def solve(x, W, b):
    eps = 1e-5
    epochs = 1000

    for i in range(epochs):
        n = W.shape[0]
        x_prev = x
        x = kaczmarz(x, W[i%n], b[i%n])

        if np.linalg.norm(x - x_prev) < eps:
            break

    print(f"Answer: x = {x}")

# Example
x = np.array([1., 2., 3.])
W = np.array([
    [3, 2, 1],
    [1, 3, 1],
    [1, 1, 5]
])
b = np.array([-6, -5, -7])

solve(x, W, b)
```
